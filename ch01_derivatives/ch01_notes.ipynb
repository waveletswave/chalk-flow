## 1.3 Total derivative along a path

Car thermometer: $\dfrac{dT}{dt} = \dfrac{\partial T}{\partial t} + v\,\dfrac{\partial T}{\partial x}$ (time effect + speed $\times$ spatial gradient).

*My notes:*

```python
dTdt = 1.0    # warming of the day [deg C per hour]
v = 60.0      # car speed [km/h]
dTdx = 0.05   # temperature gradient along the road [deg C per km]
print("driving toward the warm area:", dTdt + v * dTdx, "deg C per hour")   # 4.0
print("driving away from it:        ", dTdt - v * dTdx, "deg C per hour")   # -2.0
```

## 1.5 Method of characteristics

Characteristics in the $x$-$t$ plane: (a) constant speed, (b) time-varying speed (Example B.1), (c) speed that depends on the solution.

*My notes:*

```python
t = np.linspace(0, 2, 200)
fig, axes = plt.subplots(1, 3, figsize=(12, 3.8), sharey=True)

for x0 in np.arange(-3, 3.01, 0.5):                          # (a) U = 1
    axes[0].plot(x0 + t, t, color="C0", lw=0.8)
disp = t + 0.6 * (1 - np.cos(2 * np.pi * t)) / (2 * np.pi)   # (b) integral of U(t)
for x0 in np.arange(-3, 3.01, 0.5):
    axes[1].plot(x0 + disp, t, color="C0", lw=0.8)
for x0 in np.arange(-3, 3.01, 0.25):                         # (c) c = 1 + 0.8 exp(-x0^2)
    axes[2].plot(x0 + (1 + 0.8 * np.exp(-x0**2)) * t, t, color="C3", lw=0.8)
axes[2].axhline(1.457, color="k", ls="--", lw=1)             # breaking time t_b

titles = ["(a) U = 1", "(b) U(t) = 1 + 0.6 sin(2 pi t)", "(c) c depends on f: lines cross"]
for ax, title in zip(axes, titles):
    ax.set(xlim=(-3, 4), xlabel="x", title=title)
axes[0].set_ylabel("t")
fig.tight_layout()
fig.savefig(FIG_DIR / "ch01_characteristics.png", dpi=200)
plt.show()
```

```python
# Symbolic check of Example B.1: f = phi(x - integral of U) solves f_t + U(t) f_x = 0
x, tt, tau = sp.symbols("x t tau", real=True)
U0, U1, w = sp.symbols("U_0 U_1 omega", positive=True)
phi = sp.Function("phi")
U = lambda s: U0 + U1 * sp.sin(w * s)
f = phi(x - sp.integrate(U(tau), (tau, 0, tt)))
print(sp.simplify(sp.diff(f, tt) + U(tt) * sp.diff(f, x)))   # 0
print(f.subs(tt, 0))                                           # phi(x)
```

## 1.8 Research connection: knickpoint migration

Knickpoints travel upstream at $c = K A^m$. With $A = k_a (L - x)^2$ and $m = 1/2$ they slow down exponentially (Exercise 1.4).

*My notes:*

```python
K, ka, L = 2e-6, 1.0, 20e3   # K [1/yr], ka [-], L [m]
t_yr = np.linspace(0, 3e6, 300)
x_k = L * (1 - np.exp(-K * np.sqrt(ka) * t_yr))   # knickpoint position upstream of the outlet [m]
print(f"time to reach 1 km from the divide: {knickpoint_travel_time(K, ka, L, 1e3) / 1e6:.3f} Myr")   # 1.498

fig, ax = plt.subplots(figsize=(5, 3.5))
ax.plot(t_yr / 1e6, x_k / 1e3)
ax.axhline(L / 1e3, color="k", ls=":", label="divide")
ax.set(xlabel="time (Myr)", ylabel="distance upstream of outlet (km)")
ax.legend()
fig.tight_layout()
fig.savefig(FIG_DIR / "ch01_knickpoint.png", dpi=200)
plt.show()
```
